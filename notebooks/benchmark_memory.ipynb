##### Imports

```python
import logging
logging.basicConfig(level=logging.ERROR)
import os
import sys
from rich import print
import random
random.seed(42)
from rich.progress import track

sys.path.append("..")
from acra import procedure_to_xdl
```

```python
os.environ["CHAT_API_KEY"] = "YOUR_API_KEY"
```

#### Memory Benchmark

```python
N= 1# number of samples
PROCEDURE_PATH = "../data/datasets/Chem-RnD/"
SAMPLE = random.sample(os.listdir(PROCEDURE_PATH), N)
SAMPLE = [PROCEDURE_PATH + s for s in SAMPLE]
```

##### With XDL memory

```python
##### Simulate
EXPERIMENT_ID = "EXP_NAME"
CONTINUATION = False # if True, no initial XDL-database is provided/ or was created previously
GRAPH_TEMPLATE = None # standard graph template is used if None
ASK_AMBIGUOUS = False # request optional expert feedback
EXPAND_MEMORY = True # expand memory with new, successful procedures
USE_AMBEGUITY_AGENT = True # use ambiguitiy agent to resolve ambiguous steps
ERRORS = []
```

```python
for procedure in track(['../data/datasets/Chem-RnD/exp537.txt']):
    try:
        procedure_to_xdl(
            procedure,
            trial_dir=EXPERIMENT_ID,
            continuation=CONTINUATION,
            graph=GRAPH_TEMPLATE,
            ask_ambiguous=ASK_AMBIGUOUS,
            expand_memory=EXPAND_MEMORY,
            use_ambiguity_agent=USE_AMBEGUITY_AGENT
        )
        
        if not CONTINUATION:
            CONTINUATION = True
    except Exception as e:
        print(f"Error {e}")
        ERRORS.append((procedure, e))
        # write to file
        with open(f"../data/memory/{EXPERIMENT_ID}/errors.json", "a+") as f:
            f.write("procedure: " + procedure + ", error:" + str(e) + "\n")
```
