```python
EXPERIMENT_ID = "EXPERIMENT_NAME"
CONTINUATION = False # if True, no initial XDL-database is provided/ or was created previously
GRAPH_TEMPLATE = None # standard graph template is used if None
ASK_AMBIGUOUS = False
EXPAND_MEMORY = True
```

```python
N = ...
PROCEDURE_PATH = "../data/datasets/Chem-RnD/"
SAMPLE = random.sample(os.listdir(PROCEDURE_PATH), N)
SAMPLE = [PROCEDURE_PATH + s for s in SAMPLE]
ERRORS = []

for procedure in track(SAMPLE):
    try:
        procedure_to_xdl(
            procedure,
            trial_dir=EXPERIMENT_ID,
            continuation=CONTINUATION,
            graph=GRAPH_TEMPLATE,
            ask_ambiguous=ASK_AMBIGUOUS,
            expand_memory=EXPAND_MEMORY,
        )
        
        if not CONTINUATION:
            CONTINUATION = True
    except Exception as e:
        print(f"Error {e}")
        ERRORS.append((procedure, e))
        # write to file
        with open(f"../data/memory/{EXPERIMENT_ID}/errors.json", "a+") as f:
            f.write("procedure: " + procedure + ", error:" + str(e) + "\n")
```
